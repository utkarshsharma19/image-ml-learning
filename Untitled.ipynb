```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
# Toy data
hours_studied = [1, 2, 3, 4, 5, 6, 7]
exam_score    = [50, 55, 65, 70, 75, 82, 90]

# Put into a table (DataFrame)
data = pd.DataFrame({
    "hours_studied": hours_studied,
    "exam_score": exam_score
})

data
```

```python
# Mean and median of hours studied
mean_hours = data["hours_studied"].mean()
median_hours = data["hours_studied"].median()

# Mean and median of exam_score
mean_score = data["exam_score"].mean()
median_score = data["exam_score"].median()

print("Hours studied -> mean:", mean_hours, ", median:", median_hours)
print("Exam score    -> mean:", mean_score, ", median:", median_score)
```

```python
plt.scatter(data["hours_studied"], data["exam_score"])
plt.xlabel("Hours studied")
plt.ylabel("Exam score")
plt.title("Study time vs Exam score")
plt.show()
```

```python
# Features (X) must be 2D: shape (n_samples, n_features)
X = data[["hours_studied"]]   # double brackets -> DataFrame (2D)
y = data["exam_score"]        # target (1D)

# 1. Create the model
model = LinearRegression()

# 2. Train (fit) the model
model.fit(X, y)

# 3. Make predictions on the same data (just for demo)
y_pred = model.predict(X)

# 4. See how good it fits (Mean Squared Error)
mse = mean_squared_error(y, y_pred)

print("Model coefficient (slope):", model.coef_[0])
print("Model intercept:", model.intercept_)
print("Mean Squared Error:", mse)
```

```python
# Scatter of real data
plt.scatter(data["hours_studied"], data["exam_score"], label="Actual data")

# Regression line
x_line = np.array([[0], [8]])              # from 0 to 8 hours
y_line = model.predict(x_line)

plt.plot(x_line, y_line, label="Fitted line")

plt.xlabel("Hours studied")
plt.ylabel("Exam score")
plt.title("Study time vs Exam score with Linear Regression line")
plt.legend()
plt.show()
```

```python
new_hours = np.array([[5.5]])
predicted_score = model.predict(new_hours)[0]
print(f"If you study {new_hours[0][0]} hours, predicted score ≈ {predicted_score:.2f}")
```
